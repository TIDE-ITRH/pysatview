```python
mnf_inner = {
    "west_lon": 113.0,
    "east_lon": 114.0,
    "south_lat": -23.0,
    "north_lat": -21.0
}
smallbox=([mnf_inner['west_lon'], mnf_inner['east_lon']],
          [mnf_inner['south_lat'], mnf_inner['north_lat']])
```

```python
from pysatview.sentinel3.sentinel3_processor import Sentinel3Workflow

# Configuration parameters
layer_keys = ['sentinel3a_sst', 'sentinel3a_chl', 'sentinel3b_sst', 'sentinel3b_chl']
region = (111, -25, 114, -20)  
time_range = ("2024-10-10T00:00:00Z", "2024-10-15T00:00:00Z")

# Create workflow instance
workflow = Sentinel3Workflow()

# Run complete workflow
workflow.run_complete_workflow(layer_keys=layer_keys, region=region, time_range=time_range, smallbox=smallbox, get_all_available=True)
```

```python
pdf_files = []

png_path = workflow.processor.base_dir / 'sentinel3a' / 'sst' / 'png'

if png_path.exists() and png_path.is_dir():
    outer = [p for p in png_path.glob('*.png') if 'inner' not in p.name]
    if outer:
        latest_outer = pdf_files.append(max(outer, key=lambda p: p.stat().st_mtime))

    inner = [p for p in png_path.glob('*.png') if 'inner' in p.name]
    if inner:
        latest_inner = pdf_files.append(max(inner, key=lambda p: p.stat().st_mtime))
```

```python
pdf_files
```

```python
from PIL import Image
from datetime import datetime, timezone

# Create PDF
imgs = [Image.open(p) for p in pdf_files]
imgs = [im.convert("RGB") for im in imgs]

# first image .save() with the rest as an append list
out_path = datetime.now(timezone.utc).strftime("%Y%m%d_%H%M") + "_latest.pdf"
imgs[0].save(out_path, save_all=True, append_images=imgs[1:])
print(f"Created {out_path} with {len(imgs)} pages")  
```
