```python
from pysatview.himawari.api import HimawariAPI

himapi = HimawariAPI()
```

```python
import asyncio
from pysatview.shared.models import ProcessingRequest

async def test_himawari_workflow():
    # Step 1: Initialize the API
    api = HimawariAPI()

    # Step 2: Check if the API is available
    is_available = await api.is_available()
    print(f"API Available: {is_available}")

    # Step 3: Perform a health check
    health_status = await api.health_check()
    print(f"Health Check: {health_status}")

    # Step 4: Get system status
    system_status = await api.get_system_status()
    print(f"System Status: {system_status}")

    # Step 5: List files (e.g., .nc files)
    files = await api.list_files(satellite="Himawari-9", parameter="sst", file_type="nc")
    print(f"Files: {files}")

    # Step 6: Process data
    request = ProcessingRequest(
        start_time="2025-10-01T00:00:00",
        end_time="2025-10-01T03:00:00",
        west_lon=110.0,
        east_lon=115.0,
        south_lat=-25.0,
        north_lat=-20.0,
        time_step_hours=1
    )
    background_tasks = None  # Replace with FastAPI's BackgroundTasks if testing in FastAPI
    processing_status = await api.process_data(request, background_tasks)
    print(f"Processing Status: {processing_status}")

    # Step 7: Check task status
    task_status = await api.get_task_status(processing_status.task_id)
    print(f"Task Status: {task_status}")

# Run the test
await test_himawari_workflow()
```
